```python
import numpy as np
import pandas as pd
from IPython.display import display
import matplotlib.pyplot as plt
from scipy.stats import norm

def estimator_1(x):
    return np.mean(x)

def estimator_2(x):
    return np.median(x)

def estimator_3(x):
    weights = np.arange(1, len(x) + 1)
    weights = weights / np.sum(weights)
    return np.sum(x * weights)

def estimator_4(x):
    n = len(x)
    x_sorted = np.sort(x)
    i = np.arange(1, n + 1)
    weights = norm.pdf(norm.ppf((i - 1) / n)) - norm.pdf(norm.ppf(i / n))
    return np.sum(x_sorted * weights)

def estimator_5(x):
    return len(x) / (np.sum(1 / x))

def estimator_6(x):
    return np.cbrt(1 / len(x) * np.sum(x ** 3))

def estimator_7(x):
    return (1 / len(x) * np.sum(x ** 4)) ** (1/4)

n_values = [20, 50, 100]
k = 10_000
labels = ['1', '2', '3', '4', '5', '6', '7']

cases = [
    (0, 1),
    (0, 2),
    (4, 1),
    (4, 2)
]

for theta, sigma in cases:
    case_tables = []
    case_est_vals = []
    
    for n in n_values:
        samples = np.random.normal(loc=theta, scale=sigma, size=(k, n))

        est_vals = [
            np.apply_along_axis(estimator_1, axis=1, arr=samples),
            np.apply_along_axis(estimator_2, axis=1, arr=samples),
            np.apply_along_axis(estimator_3, axis=1, arr=samples),
            np.apply_along_axis(estimator_4, axis=1, arr=samples),
            np.apply_along_axis(estimator_5, axis=1, arr=samples),
            np.apply_along_axis(estimator_6, axis=1, arr=samples),
            np.apply_along_axis(estimator_7, axis=1, arr=samples),
        ]

        results = []
        for i, est_val in enumerate(est_vals):
            bias = np.mean(est_val) - theta
            variance = np.var(est_val)
            mse = np.mean((est_val - theta) ** 2)
            diff = np.abs(mse - bias ** 2 - variance)
            results.append({
                "Estymator": labels[i],
                "Obciążenie": bias,
                "Wariancja": variance,
                "MSE": mse,
                "Diff": diff,
            })

        df = pd.DataFrame(results)
        case_tables.append((n, df))
        case_est_vals.append((n, est_vals))

    for n, df in case_tables:
        display(df.style.set_caption(f"Tabela wyników: n = {n}, theta = {theta}, sigma = {sigma}").format(precision=4).hide(axis="index"))

    fig, axes = plt.subplots(1, len(n_values), figsize=(18, 5))
    for ax, (n, est_vals) in zip(axes, case_est_vals):
        ax.boxplot(est_vals, tick_labels=labels)
        ax.set_title(f'n = {n}, theta = {theta}, sigma = {sigma}')
        ax.set_xlabel('Estymator')
        ax.set_ylabel('Wartość estymatora')
        ax.grid(True, linestyle='--', alpha=0.6)
        ax.set_ylim(-3 + theta, 3 + theta)
    plt.tight_layout()
    plt.show()
```

Z tabel i wykresów pudełkowych wynika, że dla danych z rozkładu normalnego najlepiej radzi sobie średnia arytmetyczna (estymator 1) – ma obciążenie praktycznie równe 0 oraz najmniejszą wariancję i najniższy błąd średniokwadratowy (MSE) w każdym z testowanych przypadków.

Mediana (estymator 2) również jest nieobciążona, ale ma nieco większy rozrzut (wariancja wyższa o ok. 25–50% względem średniej). Średnia ważona (estymator 3) i estymator 4 dają wyniki poprawne, lecz wyraźnie ustępują zwykłej średniej. Z kolei estymatory 5–7 zupełnie zawodzą - w szczególności średnia harmoniczna (estymator 5) przy $\theta = 0$.

Wraz ze wzrostem $n$ (od 20 do 100) wariancja i MSE wszystkich poprawnych estymatorów zauważalnie spadają (pudełka na wykresach się zwężają), a wyższa $\sigma$ proporcjonalnie powiększa błędy. Wartości w kolumnie Diff bliskie zeru numerycznie potwierdzają tożsamość $\text{MSE} = \text{wariancja} + \text{obciążenie}^2$.

Wniosek: W przypadku rozkładu normalnego klasyczna średnia arytmetyczna jest bezkonkurencyjna. Mediana pozostaje bezpieczną alternatywą, natomiast estymatory nieliniowe są w tym zagadnieniu bezużyteczne

```python
import numpy as np
import pandas as pd
from IPython.display import display
import matplotlib.pyplot as plt
from scipy.stats import norm

def estimator_1(x):
    return np.mean(x)

def estimator_2(x):
    return np.median(x)

n = 50
k = 10_000
labels = ['1', '2']

cases = [
    (0, 1),
    (0, 2),
    (4, 1),
    (4, 2)
]

def eval(cases, distribution):
    case_tables = []
    case_est_vals = []

    for theta, sigma in cases:
        if distribution == 'normal':
            samples = np.random.normal(loc=theta, scale=sigma, size=(k, n))
        elif distribution == 'logistic':
            samples = np.random.logistic(loc=theta, scale=sigma, size=(k, n))
        elif distribution == 'cauchy':
            samples = theta + sigma * np.random.standard_cauchy(size=(k, n))

        est_vals = [
            np.apply_along_axis(estimator_1, axis=1, arr=samples),
            np.apply_along_axis(estimator_2, axis=1, arr=samples)
        ]

        results = []
        for i, est_val in enumerate(est_vals):
            bias = np.mean(est_val) - theta
            variance = np.var(est_val)
            mse = np.mean((est_val - theta) ** 2)
            diff = np.abs(mse - bias ** 2 - variance)
            results.append({
                "Estymator": labels[i],
                "Obciążenie": bias,
                "Wariancja": variance,
                "MSE": mse,
                "Diff": diff,
            })

        df = pd.DataFrame(results)
        case_tables.append((theta, sigma, df))
        case_est_vals.append((theta, sigma, est_vals))

    for theta, sigma, df in case_tables:
        display(df.style.set_caption(f"Tabela wyników: rozkład {distribution}, theta = {theta}, sigma = {sigma}").format(precision=4).hide(axis="index"))

    fig, axes = plt.subplots(1, len(cases), figsize=(20, 5))
    for ax, (theta, sigma, est_vals) in zip(axes, case_est_vals):
        ax.boxplot(est_vals, tick_labels=labels)
        ax.set_title(f'{distribution}: theta = {theta}, sigma = {sigma}')
        ax.set_xlabel('Estymator')
        ax.set_ylabel('Wartość estymatora')
        ax.grid(True, linestyle='--', alpha=0.6)
        ax.set_ylim(-3 + theta, 3 + theta)
    plt.tight_layout()
    plt.show()

eval(cases, 'normal')
eval(cases, 'logistic')
eval(cases, 'cauchy')
```

Wyniki pokazują kluczową różnicę w zachowaniu estymatorów w zależności od tego, jak często w danym rozkładzie pojawiają się wartości skrajne:

Rozkład normalny i logistyczny: Oba estymatory są nieobciążone, ale średnia osiąga mniejszą wariancję i niższy MSE niż mediana - daje dokładniejsze oszacowanie parametru. W rozkładzie logistycznym wartości bardziej oddalone od środka zdarzają się nieco częściej, dlatego przewaga średniej nad medianą jest tam już mniejsza.

Rozkład Cauchy’ego: Średnia arytmetyczna kompletnie zawodzi – regularne pojawianie się pojedynczych, gigantycznych wartości powoduje bardzo wzrost wariancji i MSE (co widać też po zniekształconych wykresach). Mediana radzi sobie świetnie – ignoruje pojedyncze ekstrema, zachowując małe obciążenie, stabilną wariancję i niski MSE.
Zmiana parametru położenia ($\theta$) nie wpływa na precyzję estymacji, natomiast wzrost parametru skali ($\sigma$) naturalnie powiększa błędy oszacowań.

Wniosek: Dla rozkładów, w których dane skupiają się blisko środka (normalny, logistyczny), efektywniejsza jest średnia. Gdy jednak w danych regularnie pojawiają się bardzo duże, skrajne obserwacje (jak w rozkładzie Cauchy'ego), jedynym sensownym wyborem jest mediana.

# Zadanie 3

```python
import numpy as np
import pandas as pd
from IPython.display import display
import matplotlib.pyplot as plt

def estimator_1(x):
    return np.mean(x)

def estimator_2(x):
    return np.median(x)

n = 50
repetitions = 10_000
theta = 0
sigma = 1

k_values = np.arange(10, 101, 10)

bias_1, var_1, mse_1 = [], [], []
bias_2, var_2, mse_2 = [], [], []

for k in k_values:
    samples_base = np.random.normal(loc=theta, scale=sigma, size=(repetitions, n - 1))

    outliers = np.full((repetitions, 1), k)
    samples = np.hstack((samples_base, outliers))

    est_1_vals = np.apply_along_axis(estimator_1, axis=1, arr=samples)
    est_2_vals = np.apply_along_axis(estimator_2, axis=1, arr=samples)

    b1 = np.mean(est_1_vals) - theta
    v1 = np.var(est_1_vals)
    m1 = np.mean((est_1_vals - theta) ** 2)
    bias_1.append(b1)
    var_1.append(v1)
    mse_1.append(m1)

    b2 = np.mean(est_2_vals) - theta
    v2 = np.var(est_2_vals)
    m2 = np.mean((est_2_vals - theta) ** 2)
    bias_2.append(b2)
    var_2.append(v2)
    mse_2.append(m2)

mean_results = pd.DataFrame({
    "k": k_values,
    "Obciążenie": bias_1,
    "Wariancja": var_1,
    "MSE": mse_1,
    "Diff": np.abs(np.array(mse_1) - np.array(bias_1) ** 2 - np.array(var_1)),
})
display(mean_results.style.set_caption("Tabela wyników dla estymatora 1 (Średnia) w zależności od k").format(precision=4).hide(axis="index"))

median_results = pd.DataFrame({
    "k": k_values,
    "Obciążenie": bias_2,
    "Wariancja": var_2,
    "MSE": mse_2,
    "Diff": np.abs(np.array(mse_2) - np.array(bias_2) ** 2 - np.array(var_2)),
})
display(median_results.style.set_caption("Tabela wyników dla estymatora 2 (Mediana) w zależności od k").format(precision=4).hide(axis="index"))

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].plot(k_values, bias_1, label='Średnia', marker='o')
axes[0].plot(k_values, bias_2, label='Mediana', marker='s')
axes[0].set_title('Obciążenie w zależności od k')
axes[0].set_xlabel('k (wartość odstająca)')
axes[0].set_ylabel('Obciążenie')
axes[0].legend()
axes[0].grid(True, linestyle='--', alpha=0.6)

axes[1].plot(k_values, var_1, label='Średnia', marker='o')
axes[1].plot(k_values, var_2, label='Mediana', marker='s')
axes[1].set_title('Wariancja w zależności od k')
axes[1].set_xlabel('k (wartość odstająca)')
axes[1].set_ylabel('Wariancja')
axes[1].legend()
axes[1].grid(True, linestyle='--', alpha=0.6)

axes[2].plot(k_values, mse_1, label='Średnia', marker='o')
axes[2].plot(k_values, mse_2, label='Mediana', marker='s')
axes[2].set_title('Błąd Średniokwadratowy (MSE)')
axes[2].set_xlabel('k (wartość odstająca)')
axes[2].set_ylabel('MSE')
axes[2].legend()
axes[2].grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

Eksperyment pokazuje, jak różnie oba estymatory reagują na obecność w danych choćby jednej błędnej lub skrajnej obserwacji $k$:

Średnia arytmetyczna: Jest bardzo wrażliwa na anomalie. Każdy wzrost wartości $k$ bezpośrednio „ciągnie” średnią w swoją stronę - jej obciążenie rośnie proporcjonalnie do wielkości błędu, a błąd średniokwadratowy (MSE) rośnie jeszcze szybciej (kwadratowo), przez co oszacowanie błyskawicznie traci jakikolwiek sens.

Mediana: Wykazuje pełną odporność. Ponieważ zależy wyłącznie od uporządkowania danych, a nie od skrajnych wartości, jej wynik pozostaje stabilny i dokładny na całym badanym zakresie (od $k=10$ do $100$), a błędy utrzymują się blisko zera.

Wniosek: Średnia zupełnie nie sprawdza się w danych, w których mogą zdarzyć się błędy pomiarowe lub wartości odstające. W takich warunkach znacznie bezpieczniejszym i stabilniejszym wyborem jest mediana.

```python
import numpy as np
import pandas as pd
from IPython.display import display
import matplotlib.pyplot as plt
from scipy.stats import norm

def estimator_1(x):
    x_mean = np.mean(x)
    return 1 / (len(x) - 1) * np.sum((x - x_mean) ** 2)

def estimator_2(x):
    x_median = np.median(x)
    return 1 / (len(x) - 1) * np.sum((x - x_median) ** 2)

def estimator_3(x):
    return np.median((x - np.mean(x)) ** 2)

n = 50
k = 10_000
labels = ['1', '2', '3']

cases = [
    (0, 1),
    (0, 2),
    (4, 1),
    (4, 2)
]

case_tables = []
case_est_vals = []

for mi, theta in cases:
    samples = np.random.normal(loc=mi, scale=np.sqrt(theta), size=(k, n))

    est_vals = [
        np.apply_along_axis(estimator_1, axis=1, arr=samples),
        np.apply_along_axis(estimator_2, axis=1, arr=samples),
        np.apply_along_axis(estimator_3, axis=1, arr=samples)
    ]

    results = []
    for i, est_val in enumerate(est_vals):
        bias = np.mean(est_val) - theta
        variance = np.var(est_val)
        mse = np.mean((est_val - theta) ** 2)
        diff = np.abs(mse - bias ** 2 - variance)
        results.append({
            "Estymator": labels[i],
            "Obciążenie": bias,
            "Wariancja": variance,
            "MSE": mse,
            "Diff": diff,
        })

    df = pd.DataFrame(results)
    case_tables.append((mi, theta, df))
    case_est_vals.append((mi, theta, est_vals))

for mi, theta, df in case_tables:
    display(df.style.set_caption(f"Tabela wyników estymatorów wariancji dla mean = {mi}, theta = {theta}").format(precision=4).hide(axis="index"))

fig, axes = plt.subplots(1, len(cases), figsize=(20, 5))
for ax, (mi, theta, est_vals) in zip(axes, case_est_vals):
    ax.boxplot(est_vals, tick_labels=labels)
    ax.set_title(f'mean = {mi}, theta = {theta}')
    ax.set_xlabel('Estymator')
    ax.set_ylabel('Wartość estymatora')
    ax.grid(True, linestyle='--', alpha=0.6)
    ax.set_ylim(-3 + theta, 3 + theta)
plt.tight_layout()
plt.show()
```

W badaniu estymacji wariancji dla rozkładu normalnego widać wyraźne różnice między testowanymi podejściami:

Estymator 1: Radzi sobie najlepiej. Daje wyniki praktycznie nieobciążone i ma najmniejszy błąd średniokwadratowy (MSE), co czyni go najbardziej precyzyjnym.

Estymator 2: Działa poprawnie i trafia w okolice prawdziwej wartości, ale zastąpienie średniej medianą wprowadza lekkie zawyżenie wyniku (dodatnie obciążenie) oraz nieco większy rozrzut.

Estymator 3: Bardzo mocno zaniża faktyczną wariancję, przez co jego błąd (MSE) jest największy. Konstrukcja oparta na samej medianie odchyleń z natury szacuje inną wielkość i bez odpowiedniego współczynnika skalującego systematycznie podaje zbyt małe wartości.

Dodatkowo zmiana średniej ($\mu = 0$ na $\mu = 4$) w żaden sposób nie wpływa na estymację wariancji, natomiast wzrost samej wariancji (z 1 na 2) proporcjonalnie zwiększa błędy wszystkich metod.

Wniosek: Dla czystych danych z rozkładu normalnego standardowy estymator wariancji $S^2$ pozostaje bezkonkurencyjny. Metody oparte na medianie odchyleń wymagają dodatkowego przeskalowania, aby dawały wiarygodny poziom wariancji.
